````python
# Verify that the API key is loaded correctly.
# If you're using a different provider, replace with the appropriate environment variable.
# Note: Ollama does not require an API key.

import os
openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key detected, starting with: {openai_api_key[:8]}...")
else:
    print("OpenAI API Key is missing. Please refer to the troubleshooting guide in the setup folder.")
````

````python
# Now, import the OpenAI client library.
# If you encounter an ImportError, please check the troubleshooting guide in the Setup folder.

from openai import OpenAI
````

````python
# === LLM SETUP (edit only this cell to change provider/model) ===
# Reads from project-root .env; defaults to Ollama.

openai = OpenAI(
    base_url=os.getenv("LOCAL_LLM_BASE_URL", "http://localhost:11434/v1"),
    api_key=os.getenv("LOCAL_LLM_API_KEY", "ollama"),
)
model_name = os.getenv("LOCAL_LLM_MODEL", "llama3.2:latest")
print(f"Using model: {model_name}")
````

````python
# Prepare a list of messages in the OpenAI format with a different example question.

messages = [{"role": "user", "content": "Can you explain the basic principles of photosynthesis?"}]
````

````python
# Uses model_name from the LLM SETUP cell above.

response = openai.chat.completions.create(
    model=model_name,
    messages=messages
)

print(response.choices[0].message.content)
````

````python
# Now, let's request the model to generate a challenging IQ test question.

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
````

````python
# Same shared model from the LLM SETUP cell.

response = openai.chat.completions.create(
    model=model_name,
    messages=messages
)

question = response.choices[0].message.content

print(question)
````

````python
# Create a new messages list with the generated question as the user input.

messages = [{"role": "user", "content": question}]
````

````python
# Same shared model from the LLM SETUP cell.

response = openai.chat.completions.create(
    model=model_name,
    messages=messages
)

answer = response.choices[0].message.content
print(answer)
````

````python
from IPython.display import Markdown, display

# Display the response as formatted Markdown for better readability
display(Markdown(answer))
````
